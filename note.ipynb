```python
from typing import Optional, List, Dict, Any
import os
from datetime import datetime, date  # date型もインポート
import re
def _clean_date_columns(data: Any, schema: Optional[List[Any]] = None) -> None:
        if not isinstance(data, list):
            return

        # スキーマがある場合だけ「DATE型」のカラム名を特定する
        date_type_cols = set()
        if schema:
            date_type_cols = {f.name for f in schema if getattr(f, "field_type", "").upper() == 'DATE'}

        for row in data:
            for key, val in row.items():
                if val is None:
                    continue

                # 中身が datetime/date 型なら、列名に関わらず文字列に変換
                if isinstance(val, (datetime, date)):
                    dt_val = val.replace(tzinfo=None) if isinstance(val, datetime) else val
                    
                    # スキーマでDATE型と指定されていれば YYYY-MM-DD
                    if key in date_type_cols:
                        row[key] = dt_val.isoformat()[:10]
                    else:
                        # それ以外（DATETIME）やスキーマがない場合はフルISO
                        row[key] = dt_val.isoformat()

                # 文字列の場合の整形ロジック
                elif isinstance(val, str):
                    nums = re.findall(r'\d+', val)
                    if len(nums) == 3:
                        y, m, d = nums
                        row[key] = f"{y}-{m.zfill(2)}-{d.zfill(2)}"
```
